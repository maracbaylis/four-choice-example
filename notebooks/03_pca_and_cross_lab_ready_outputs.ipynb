## Future Multi-Lab Pattern

For multiple labs, run the pipeline once per lab or submission. Then combine the resulting `animal_level_outputs.csv` files.

```python
# Example with one lab output; add more CSV paths as more labs submit data.
animal_csvs = [OUTDIR / "tables" / "animal_level_outputs.csv"]
combined_animals = combine_animal_tables(animal_csvs)
combined_animals[[c for c in ["animal_id", "lab_source", "experiment_cohort", "group", "sex", "TTC_in_discrimination", "TTC_in_reversal"] if c in combined_animals.columns]]
```

Recommended final workflow:

1. Run QC and PCA within each lab submission.
2. Combine `animal_level_outputs.csv` across labs.
3. Run a pooled PCA with lab/source labels retained.
4. Inspect whether PCA structure tracks treatment, behavioral phenotype, or lab/site effects.
